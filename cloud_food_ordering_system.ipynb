```python
# Cloud-Based Food Ordering System
# Run directly in Google Colab

import sqlite3

# Create database
conn = sqlite3.connect("food_ordering.db")
cursor = conn.cursor()

# Create food table
cursor.execute("""
CREATE TABLE IF NOT EXISTS food (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    price REAL NOT NULL
)
""")

# Create orders table
cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY AUTOINCREMENT,
    customer_name TEXT NOT NULL,
    food_name TEXT NOT NULL,
    quantity INTEGER NOT NULL,
    total REAL NOT NULL
)
""")

# Add food items only if table is empty
cursor.execute("SELECT COUNT(*) FROM food")
if cursor.fetchone()[0] == 0:
    food_items = [
        (1, "Pizza", 150),
        (2, "Burger", 100),
        (3, "Fried Rice", 120),
        (4, "Biryani", 180),
        (5, "Dosa", 60)
    ]

    cursor.executemany(
        "INSERT INTO food (id, name, price) VALUES (?, ?, ?)",
        food_items
    )
    conn.commit()


# Display food menu
def show_menu():
    print("\n========== FOOD MENU ==========")
    cursor.execute("SELECT * FROM food")
    foods = cursor.fetchall()

    for food in foods:
        print(f"{food[0]}. {food[1]} - ₹{food[2]:.2f}")

    print("===============================")


# Place order
def place_order():
    show_menu()

    customer = input("\nEnter customer name: ")

    try:
        food_id = int(input("Enter food ID: "))
        quantity = int(input("Enter quantity: "))

        cursor.execute(
            "SELECT name, price FROM food WHERE id = ?",
            (food_id,)
        )

        food = cursor.fetchone()

        if food is None:
            print("Invalid food ID!")
            return

        if quantity <= 0:
            print("Quantity must be greater than 0!")
            return

        food_name = food[0]
        price = food[1]
        total = price * quantity

        cursor.execute("""
        INSERT INTO orders
        (customer_name, food_name, quantity, total)
        VALUES (?, ?, ?, ?)
        """, (customer, food_name, quantity, total))

        conn.commit()

        print("\nOrder placed successfully!")
        print("Customer :", customer)
        print("Food     :", food_name)
        print("Quantity :", quantity)
        print("Total    : ₹", total)

    except ValueError:
        print("Please enter valid numbers!")


# View orders
def view_orders():
    print("\n========== ALL ORDERS ==========")

    cursor.execute("SELECT * FROM orders")
    orders = cursor.fetchall()

    if not orders:
        print("No orders found.")
    else:
        for order in orders:
            print(
                f"Order ID: {order[0]} | "
                f"Customer: {order[1]} | "
                f"Food: {order[2]} | "
                f"Quantity: {order[3]} | "
                f"Total: ₹{order[4]:.2f}"
            )

    print("===============================")


# Main program
while True:

    print("\n===== CLOUD-BASED FOOD ORDERING SYSTEM =====")
    print("1. View Food Menu")
    print("2. Place Food Order")
    print("3. View All Orders")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        show_menu()

    elif choice == "2":
        place_order()

    elif choice == "3":
        view_orders()

    elif choice == "4":
        print("\nThank you for using Food Ordering System!")
        break

    else:
        print("Invalid choice! Please try again.")

conn.close()
```
